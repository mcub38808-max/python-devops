```python
class Deployment:
    """
    Manages the state and version history of a software deployment.
    """

    def __init__(self, service_name: str, environment: str):
        self.service_name = service_name
        self.environment = environment
        self.status = 'pending'
        self.version = None
        self._history = []

    def deploy(self, new_version: str):
        self.status = 'deployed'
        
        if self.version:
            self._history.append(self.version)
            self.version = new_version
        else:
            self.version = new_version

    def rollback(self) -> bool:
        if not self._history:
            return False
        else:
            self.version = self._history.pop()
            self.status = 'rolled_back'
            return True

    def check_status(self) -> dict:
        return {
            "service_name": self.service_name,
            "environment": self.environment,
            "status": self.status,
            "version": self.version
        }

nginx_deployment = Deployment('nginx', 'production')
```

```python
type(nginx_deployment)
```

```python
nginx_deployment.check_status()
```

```python
nginx_deployment.deploy('1.0.0')
```

```python
nginx_deployment.check_status()
```

```python
nginx_deployment.deploy('1.0.1')
```

```python
nginx_deployment.check_status()
```

```python
nginx_deployment.deploy('1.1.1')
nginx_deployment.check_status()
```

```python
nginx_deployment.rollback()
nginx_deployment.check_status()
```
