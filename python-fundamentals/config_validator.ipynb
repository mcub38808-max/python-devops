```python
def validate_config(config: dict) -> bool:
    """
    Validates a configuration dictionary against a set of rules.

    A valid configuration must have:
    - Required keys: 'service_name', 'env', 'port'.
    - 'env' must be one of 'dev', 'staging', 'prod'.
    - 'service_name' must be a non-empty string.
    - 'port' must be an integer between 1 and 65535.

    Args:
        config: A dictionary containing configuration parameters.

    Returns:
        True if the configuration is valid, False otherwise.
    """
    # TODO: Implement the validation logic here.
    # 1. Check if all required keys exist.
    
    valid_config = False
    required_keys = ["service_name", "env", "port"]
    valid_envs = ["dev","staging","prod"]
    
    missing_key  = list(required_keys - config.keys())
    
    if missing_key:
        return False
    elif config["env"] not in valid_envs:   # 2. Check if the 'env' value is one of the allowed environments.
        return False
    #elif not (config["service_name"] and type(config["service_name"]) == str):    # 3. Check if 'service_name' is a non-empty string.
    elif not (config["service_name"] and isinstance(config["service_name"],str)):    # 3. Check if 'service_name' is a non-empty string.
        return False
    #elif not (type(config["port"]) == int and config["port"] in range(1,65536)):   # 4. Check if 'port' is an integer and within the valid range.
    elif not (isinstance(config["port"], int) and config["port"] in range(1,65536)):   # 4. Check if 'port' is an integer and within the valid range.
        return False
    else:
        return True
```
